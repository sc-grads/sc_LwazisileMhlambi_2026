```python
seq = [1,2,3,4,5]
```

```python
def times2(var):
    return var*2
```

```python
map(times2, seq)
```

```python
list(map(times2, seq))
```

```python
#Lambda Expression
t = lambda var:var*2
```

```python
t(6)
```

```python
list(map(lambda num: num*3, seq))
```

```python
#Filter Function

#Filtered for even numbers
list(filter(lambda num: num%2==0, seq))
```
