```python
seq = [1,2,3,4,5]
```

```python
for item in seq:
    print(item)
```

```python
i = 1

while i < 5:
    print('i is: {}'.format(i))
    i = i+1
```

```python
for x in seq:
    print(x)
```

```python
#Range function
list(range(10))
```

```python
for x in range(0,5):
    print(x)
```

```python
#List comprehension
x = [1,2,3,4]
```

```python
out =[]

for num in x:
    out.append(num**2)
```

```python
print(out)
```

```python
#What you want for which list
[num**2 for num in x]
```
