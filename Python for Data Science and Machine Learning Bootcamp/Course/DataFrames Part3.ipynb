```python
import numpy as np
import pandas as pd
from numpy.random import randn
```

```python
#Index Levels
outside = ['G1','G1','G1','G2','G2','G2']
inside = [1,2,3,1,2,3]
hier_index = list(zip(outside, inside))
hier_index = pd.MultiIndex.from_tuples(hier_index)
```

```python
outside
```

```python
inside
```

```python
hier_index
```

```python
df = pd.DataFrame(randn(6,2),hier_index,['A','B'])
```

```python
#Multi-level index
df
```

```python
#Calling from multi-level index
df.loc['G1'].loc[1]
```

```python
#Labelling index columns
df.index.names = ['Groups', 'Num']
```

```python
df
```

```python
df.loc['G2'].loc[2].loc['B']
```

```python
#Returns Cross Section from a Data frame
df.xs
```

```python
df.xs('G1')
```

```python
#Can specify which level of an index you want
df.xs(1,level='Num')
```
